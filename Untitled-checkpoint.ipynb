```python
import os#作业

os.makedirs(os.path.join('..', 'data'), exist_ok=True)
data_file = os.path.join('..', 'data', 'house_tiny.csv')
with open(data_file, 'w') as f:
    f.write('BLG,AL,TES,IG\n')
    f.write('NA,Pave,1,NA\n')
    f.write('10,NA,1,1\n')
    f.write('9,NA,1,NA\n')
    f.write('NA,NA,2,1\n')
```

```python
import pandas as pd

data = pd.read_csv(data_file)
print(data)
```

```python
inputs = data.iloc[:, :]
numeric_mean = inputs.select_dtypes(include='number').mean()#mean取平均值
inputs = inputs.fillna(numeric_mean)
inputs = pd.get_dummies(inputs, dummy_na=True).astype(int)
print(inputs)
```

```python
import torch
X = torch.tensor(inputs.to_numpy(dtype='float32'))
X
```
